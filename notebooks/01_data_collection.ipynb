```python
from pathlib import Path
import pandas as pd

# Use Path to build Windows-safe paths and avoid backslash-escape issues
base = Path(r"C:\Users\LENOVO\Desktop\Kavi docu\jobs\f1-business-analytics")

races = pd.read_csv(base / "data" / "raw" / "races.csv")
drivers = pd.read_csv(base / "data" / "raw" / "drivers.csv")
constructors = pd.read_csv(base / "data" / "raw" / "constructors.csv")
results = pd.read_csv(base / "data" / "raw" / "results.csv")
pit_stops = pd.read_csv(base / "data" / "raw" / "pit_stops.csv")

print(races.shape, drivers.shape, constructors.shape, results.shape, pit_stops.shape)
```

```python
def inspect_df(df, name):
    print(f"\n{name}")
    print("-"*40)
    print(df.info())
    print("\nNull values:\n", df.isnull().sum())
    print("\nDuplicates:", df.duplicated().sum())

inspect_df(races, "RACES")
inspect_df(drivers, "DRIVERS")
inspect_df(constructors, "CONSTRUCTORS")
inspect_df(results, "RESULTS")
inspect_df(pit_stops, "PIT STOPS")
```

```python
races.drop_duplicates(inplace=True)
drivers.drop_duplicates(inplace=True)
constructors.drop_duplicates(inplace=True)
results.drop_duplicates(inplace=True)
pit_stops.drop_duplicates(inplace=True)
```

```python
races.isnull().sum()
drivers.isnull().sum()
constructors.isnull().sum() 
results.isnull().sum()
pit_stops.isnull().sum()
```

```python
drivers['nationality'].fillna('Unknown', inplace=True)
```

```python
constructors['nationality'].fillna('Unknown', inplace=True)
```

```python
results.isnull().sum()
```

```python
results['points'].fillna(0, inplace=True)
```

```python
results['dnf'] = results['position'].isnull().astype(int)
```

```python
results['position'].fillna(0, inplace=True)
```

```python
races['date'] = pd.to_datetime(races['date'])
```

```python
num_cols = ['grid', 'position', 'points', 'laps']
results[num_cols] = results[num_cols].apply(pd.to_numeric)
```

```python
import numpy as np

results.replace("\\N", np.nan, inplace=True)
```

```python
num_cols = ['grid', 'position', 'points', 'laps']

results[num_cols] = results[num_cols].apply(
    pd.to_numeric, errors='coerce'
)
```

```python
results['points'].fillna(0, inplace=True)
```

```python
results['dnf'] = results['position'].isna().astype(int)
results['position'].fillna(0, inplace=True)
```

```python
results[num_cols].isnull().sum()
results[num_cols].dtypes
```

```python
for df_ in [races, drivers, constructors, results]:
    df_.replace("\\N", np.nan, inplace=True)
```

```python

results.describe()
```

```python
results = results[results['laps'] >= 0]
results = results[results['grid'] >= 0]
```

```python
df = results.merge(races, on='raceId', how='left')
df = df.merge(drivers, on='driverId', how='left')
df = df.merge(constructors, on='constructorId', how='left')
```

```python
inspect_df(df, "FINAL DATASET")
df.head()
```

```python
df['driver_name'] = df['forename'] + ' ' + df['surname']
df['season'] = df['year']
df['grid_vs_finish'] = df['grid'] - df['position']
```

```python
df.to_csv("../data/processed/f1_clean_analytics_dataset.csv", index=False)
```
